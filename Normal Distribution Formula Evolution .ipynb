```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x0=np.linspace(-4,4,10000)
y0=np.e**(-(x0**2))
y1=(1/np.sqrt(np.pi))*(np.e**(-(x0**2)))

x2=np.linspace(0,200,10000)
y2= ( 1 / 
          (15* np.sqrt( 2*np.pi ) ) * 
                                       (np.e**(- 
                                                  (( x2 - 100)
                                                            / (2*15))**2 ) ) )
std3=0.5
std4=0.25
std5=0.25
y3= (1 / np.sqrt(np.pi)) *  np.e**-((1/2 *(x0/std3))**2)
y4= (1 / np.sqrt(np.pi)) *  np.e**-((1/2 *(x0/std4))**2)
y5= ( 1 / std5 * np.sqrt( 2* np.pi) ) *  np.e**-((1/2 *(x0/std5))**2)


fig,ax=plt.subplots(2,3,figsize=(12,7))
ax[0][0].plot(x0,y0,color="red")
ax[0][1].plot(x0,y1,color="green")
ax[0][2].plot(x0,y3,color="indigo")

ax[1][0].plot(x0,y4,color="cyan")
ax[1][1].plot(x0,y5,color="purple")
ax[1][2].plot(x2,y2,color="brown")

ax[0][0].set_title("$e^{-X^2}$ Area is $\\sqrt{\\pi}$",color="red")
ax[0][1].set_title("$\\frac{1}{\\sqrt{\\pi}} \\cdot e^{-X^2}$ Area is 1",color="green")
ax[1][2].set_title("$ f(x) = \\frac{1}{\\sigma  \\cdot \\sqrt{2 \\cdot \\pi} }  \\cdot e^{ -( \\frac{x - \\mu }{2 \\cdot \\sigma} )^2}$ shift+spread",color="brown")
ax[0][2].set_title("$\\frac{1}{\\sqrt{\\pi}} \\cdot e^{-\\frac{1}{2}(\\frac{x}{\\sigma})^2} $ Area is $\\sigma \\cdot \\sqrt{2}$ $\\sigma=0.5$",color="magenta")
ax[1][0].set_title("$\\frac{1}{\\sqrt{\\pi}} \\cdot e^{-\\frac{1}{2}(\\frac{x}{\\sigma})^2} $ Area is $\\sigma \\cdot \\sqrt{2}$ $\\sigma=0.25$",color="teal")
ax[1][1].set_title("$\\frac{1}{\\sigma \\cdot \\sqrt{2\\pi}} \\cdot e^{-\\frac{1}{2}(\\frac{x}{\\sigma})^2} $ Area is 1",color="purple")
# plt.savefig("normal formula evolution.pdf")
plt.subplots_adjust(left=0.1, right=0.9, top=3, bottom=2)
for i in range(2):
    for j in range(3):
        ax[i,j].axis("off")
plt.show()
```
