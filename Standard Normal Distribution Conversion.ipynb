```python
import numpy as np
import scipy
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
size1=10000000
mu=65                      # update is needed -------------------------
std=20                     # update is needed -------------------------
x=40                       # update is needed -------------------------
zscore=(x-mu)/std
colors=["green","brown"]

fig,ax=plt.subplots(2,1,figsize=(12,6))
x1=np.random.normal(loc=mu,scale=std,size=size1)
x2=np.random.randn(size1)
sns.kdeplot(x1,ax=ax[0],color=colors[1])
sns.kdeplot(x2,ax=ax[1],color=colors[1])

ax[0].axvline(x,0,.2,lw=4,ls="--",color=colors[0])
ax[0].axvline(mu,0,.95,lw=0.5,ls="--",color=colors[0])

ax[1].axvline(zscore,0,0.2,lw=4,ls="-.",color=colors[1])
ax[1].axvline(0,0,0.95,lw=0.5,ls="-.",color=colors[1])
lst1=[mu,0]
lst2=[x,zscore]

# fill-between ---------areas
xf1=np.linspace(0,x,100)
xf2=np.linspace(-4,zscore,100)
yf1=stats.norm.pdf(xf1,loc=mu,scale=std)
yf2=stats.norm.pdf(xf2)
fillx=[xf1,xf2]
filly=[yf1,yf2]

for i in range(2):
    # ax[i].axis("off")
    ax[i].set_title("standardization conversion mean is :{}   $x${}".format(lst1[i],lst2[i]),x=0.1,color=colors[i])
    ax[i].fill_between(fillx[i],filly[i],color=colors[i])

plt.show()
```
