```python
import scipy 
from scipy import stats
```

```python
zsc1=2.35
zsc2=-1.25
zsc3=2.00

zsc41=0.35  # samller value
zsc42=1.85  # larger value 

lis=[zsc1,zsc2,zsc3,zsc41,zsc42]
for i in lis:
    print("------------------------------------------------------------------------")
    print("zscore :  {} probability is : {}".format(i,stats.norm.cdf(i)))
    print("------------------------------------------------------------------------")

print("zscore between {} and {} |  | probability is {}".format(zsc41,zsc42,stats.norm.cdf(zsc42)-stats.norm.cdf(zsc41)))
```

```python
zsc1=2.35
zsc2=-1.25
zsc3=2.00

zsc41=0.35  # samller value
zsc42=1.85  # larger value 

lis=[zsc1,zsc2,zsc3,zsc41,zsc42]
for i in lis:
    print("------------------------------------------------------------------------")
    print("zscore :  {} probability is : {}".format(i,1-stats.norm.sf(i)))
    print("------------------------------------------------------------------------")

print("zscore between {} and {} |  | probability is {}".format(zsc41,zsc42,(stats.norm.sf(zsc41)-stats.norm.sf(zsc42))))
```

```python
prob1=0.9906132944651614
prob2=0.10564977366685535
prob3=0.9772498680518208

prob41=0.6368306511756191  # samller value
prob42=0.9678432252043863  # larger value 

lis=[prob1,prob2,prob3,prob41,prob42]
for i in lis:
    print("------------------------------------------------------------------------")
    print("probability :  {}  zscore is : {}".format(i,stats.norm.ppf(i)))
    print("------------------------------------------------------------------------")

print("probabiity  {} and {} |  | zscore is {} and {}".format(prob41,prob42,stats.norm.ppf(prob42),stats.norm.ppf(prob41)))
```
