## quiz_01

<li>IQS are normally distributed with a mean of 100 and a standard deviation of 15.
           <br>What percentage of people have an IQ lower than 85?
        <span style="background-color:indigo"> Answer:         15.865%</span>
    <br>What percentage of peopel have an IQ greater than 130?
        <span style="background-color:indigo"> Answer:         2.275%</span>
        <br>What is the probability that a random person has an IQ less than 115?
        <span style="background-color:indigo">84.13%</span>
        <br>What is the probability that a random IQ is greater than 131.5?
        <span style="background-color:indigo">1.78%</span>
        <br>What is the probability of random IQ between 90 and 110?
        <span style="background-color:indigo">49%</span></li>

```python
mu       =100
std      =15

x1_left  =85
zscore1_left=(x1_left-mu)/std
print("IQS are normally distributed with a mean of 100 and a standard deviation of 15")
print("What percentage of people have an IQ lower than 85?                    ans1 :{:.3f}%".\
    format( stats.norm.cdf(zscore1_left) *100))

x1_right  =130
zscore1_right=(x1_right-mu)/std
print("What percentage of peopel have an IQ greater than 130?                 ans2 :{:.3f}%".\
    format( stats.norm.sf(zscore1_right) *100))

x2_left  =115
zscore2_left=(x2_left-mu)/std
print("What is the probability that a random person has an IQ less than 115?  ans3 :{:.3f}%".\
    format( stats.norm.cdf(zscore2_left) *100))

x2_right  =131.5
zscore2_right=(x2_right-mu)/std 
print("What is the probability that a random IQ is greater than 131.5?        ans4 :{:.3f}%".\
    format( stats.norm.sf(zscore2_right) *100))

x_b_1_left  =110 # larger
x_b_2_left  =90
zscore1_btw=(x_b_1_left-mu)/std
zscore2_btw=(x_b_2_left-mu)/std
print("What is the probability of random IQ between 90 and 110?               ans5 :{:.3f}%".\
    format( stats.norm.cdf(zscore1_btw) *100 - stats.norm.cdf(zscore2_btw)*100))
```

## quiz_02

 <li>The human body temperature is normally distributed with a mean of 98.2 degrees and a standard deviation of 0.62 degrees. 
        <br>If a fever is defined to be greater than 100.6 degrees, what percentage of normal healthy people have a "fever"?<span style="background-color:indigo"> 0.0054%</span></li>

```python
mean    =98.2
std     =0.62
x       =100.6
zscore  =(x-mean)/std
prob1   = 1-stats.norm.cdf(zscore)
prob2   = stats.norm.sf(zscore)
print("The human body temperature is normally distributed with a mean of 98.2 degrees and a standard deviation of 0.62 degrees.")
print("If a fever is defined to be greater than 100.6 degrees, what percentage of normal healthy people have a 'fever'?mtheod1 {:.6f}%".format(prob1*100) )
print("If a fever is defined to be greater than 100.6 degrees, what percentage of normal healthy people have a 'fever'?method2 {:.6f}%".format(prob2*100) )
```

## quiz_03

 <li>A standardized test score is normally distributed with a mean score of 998 and standarad deviation of 202. <br>If the minimum score to get into a university is 1100. what percentage of people don't get in? <span style="background-color:indigo">zscore=0.53 percentage 69.32%</span>
    <br>If the admission requirement is changed so that you will be admitted 
        <br>&emsp;&emsp;only when you are in the top 40% of the score, what would be the minimum score required to get in?<span style="background-color:indigo">1049</span></li>

```python
mean   =998
std    =202
x      =1100
zscore =(x-mean)/std
prob1  =stats.norm.cdf(zscore)
print("A standardized test score is normally distributed with a mean score of 998 and standarad deviation of 202.")
print("If the minimum score to get into a university is 1100. what percentage of people don't get in? ans:{:.3f}%".format(prob1*100))

zscore1=stats.norm.ppf(0.4) # ppf returns the area to the left therefore mirror the the result to the right of the tail is needed
x1=std*np.abs(zscore1)+mean
print("If the admission requirement is changed so that you will be admitted ,\
        \n only when you are in the top 40% of the score, what would be the minimum score required to get in? \
        \n ans:{:,}".format(int(x1)))
```

## quiz_04

<li>Replacement times for DVD players are normally distributed with a mean of 8.2 years and a standard deviation of 1.1 years.
        <br>What's the probability that a random DVD player will need to be replaced in less than 5 years?<span style="background-color:indigo">0.18%</span>
    <br>If you want to offer a warranty so that only 1% of the DVD players get replaced. How long should the warranty be?<span style="background-color:indigo">5.64 years</span></li>

```python
mean = 8.2
std  = 1.1
x    =  5  # less tnan 5 year area to the left

zscore =(x-mean)/std
prob   =stats.norm.cdf(zscore)
print("Replacement times for DVD players are normally distributed with a mean of 8.2 years and a standard deviation of 1.1 years.")
print("What's the probability that a random DVD player will need to be replaced in less than 5 years? ans {:.2f}%".\
       format(prob*100) )
prob1   =0.01 #1%
zscroe1 =stats.norm.ppf(prob1)
x1      = (zscroe1*std)+mean
print("What's the probability that a random DVD player will need to be replaced in less than 5 years? ans {:.2f} years".\
       format(x1 ) )
```

## quiz_05

<li>Body temperature is normally distributed with a mean of 98.6 dgrees ferhenhit and a standarad deviation of 0.73 degrees
        <br>What is the probability of having a normal body temperature less than 96.9 degrees fahrenheit  ?
        <span style="background-color:indigo"> Answer:         0.9935%</span>
        <br>What is the probability of having a normal body temperature greater than 100 degrees fahrenheit ?
        <span style="background-color:indigo"> Answer:         2.75677%</span>
        <br>What is the probability of having a normal body temperature ranging between 98 and 99 degrees fahrenheit ?
        <span style="background-color:indigo"> Answer:         50.2573%</span>
        <br>What is the probability of having a normal body temperature less tnan 97.6 or greater than 99.6 degrees fahrenheit ?
        <span style="background-color:indigo"> Answer:         17.07297%</span>
</li>

```python
mean=98.6
std =0.73

x1  =96.9 # less than 96.9
zscore1 =(x1-mean)/std
prob1   =stats.norm.cdf(zscore1)*100
print("Body temperature is normally distributed with a mean of 98.6 dgrees ferhenhit and a standarad deviation of 0.73 degrees")
print("What is the probability of having a normal body temperature less than 96.9 degrees fahrenheit  ? ans:  {:.5f}%".format(prob1))

x2  =100 # greater than 100
zscore2 =(x2-mean)/std
prob2   =stats.norm.sf(zscore2)*100
print("What is the probability of having a normal body temperature greater than 100 degrees fahrenheit ? ans:  {:.5f}%".format(prob2))

x_bwt_1=98
x_bwt_2=99

zscore_bwt_1=(x_bwt_1-mean)/std
zscore_bwt_2=(x_bwt_2-mean)/std
prob3   = (stats.norm.cdf(zscore_bwt_2)-stats.norm.cdf(zscore_bwt_1)) *100
print("What is the probability of having a normal body temperature ranging between 98 and 99 degrees fahrenheit ? ans:  {:.5f}%".format(prob3))

x_l_t1 = 97.6 #less than area to the left
x_g_t1 = 99.6 #greater than are on right

zscore_l_t1 = (x_l_t1 - mean) / std
zscore_g_t1 = (x_g_t1 - mean) / std
prob4    = (stats.norm.cdf(zscore_l_t1) + stats.norm.sf(zscore_g_t1)) *100
print("What is the probability of having a normal body temperature less tnan 97.6 or greater than 99.6 degrees fahrenheit ? ans:  {:.5f}%".format(prob4))
```
